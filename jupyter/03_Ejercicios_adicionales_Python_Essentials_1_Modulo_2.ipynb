## Ejercicio 3. Precio con IVA

Pide el precio de un producto antes de impuestos. Calcula el importe del IVA suponiendo un tipo del 21 % y el precio final.

**Ejemplo orientativo:** si el precio inicial es `50`, el IVA será `10.5` y el precio final será `60.5`.

**Requisitos:**
- Utiliza variables para el precio inicial, el IVA y el precio final.
- Convierte el dato introducido a un número decimal.
- Muestra los importes con dos decimales.


````python
precio = float(input("Precio: "))
iva = precio * 0.21
precio_final = precio + iva
print(f'Precio final: {precio_final:.2f}')
````

## Ejercicio 4. Conversor de minutos

Solicita una cantidad total de minutos y conviértela a horas y minutos.

**Ejemplo:** para `135` minutos, el programa debe mostrar `2 horas y 15 minutos`.

**Pista:** puedes utilizar la división entera (`//`) y el resto (`%`).


````python
minutos = int(input("Minutos: "))
horas = minutos // 60
minutos = minutos % 60
print(f'{horas}:{minutos}')
````

## Ejercicio 5. Operaciones con dos números

Solicita dos números enteros y muestra:
- La suma.
- La resta del primero menos el segundo.
- El producto.
- La división del primero entre el segundo.
- El resto de la división entera.

**Importante:** en este ejercicio, puedes suponer que el segundo número no es cero.


````python
numero1 = int(input("Número1: "))
numero2 = int(input("Número2: "))
print(numero1 + numero2)
print(numero1 - numero2)
print(numero1 * numero2)
print(numero1 // numero2) # división entera
print(numero1 % numero2)
````

## Ejercicio 6. Área y perímetro de un rectángulo

Solicita la base y la altura de un rectángulo. Calcula y muestra su área y su perímetro.

**Fórmulas:**
- Área = base × altura
- Perímetro = 2 × (base + altura)

Introduce las medidas como números decimales y muestra los resultados con dos cifras decimales.


````python
base = float(input("Base: "))
altura = float(input("Altura: "))
perimetro = 2 * (base + altura)
print(f'Perímetro = {perimetro:.2f}')
````

## Ejercicio 7. Intercambio de valores

Crea dos variables, `a` y `b`, con los valores `12` y `35`. Muestra sus valores iniciales, intercámbialos y vuelve a mostrarlos.

**Resultado esperado:**
```text
Antes del intercambio: a = 12, b = 35
Después del intercambio: a = 35, b = 12
```

Resuelve el intercambio utilizando una variable auxiliar.


````python
a = 12
b = 35
print(f'Valores iniciales: a = {a}, b = {b}')
# invento: variable auxiliar
aux = a # 12
a = b # 35
b = aux # 35
print(f'Valores iniciales: a = {a}, b = {b}')
````

## Ejercicio 8. Segundos a horas, minutos y segundos

Solicita una cantidad de segundos y conviértela a horas, minutos y segundos.

**Ejemplo:** para `3672` segundos, el resultado debe ser `1 hora, 1 minuto y 12 segundos`.

**Pista:** utiliza `//` y `%` para obtener cada componente.


````python
segundos = int(input("Segundos totales: "))
minutos = segundos // 60
segundos = segundos % 60
horas = minutos // 60
minutos = minutos % 60
print(f'{horas}:{minutos}:{segundos}')
````
