## Ejercicio 4. Operadores de comparación

Declara `edad = 17` y `edad_minima = 18`. Muestra los resultados de `==`, `!=`, `<`, `>`, `<=` y `>=` al comparar ambas variables.

```python
# Escribe aquí tu solución
edad = 17
edad_minima = 18
print(edad == edad_minima)
print(edad != edad_minima)  
print(edad > edad_minima) 
print(edad < edad_minima)
print(edad >= edad_minima)
print(edad <= edad_minima)
```
