## Ejercicio 1. Contar del 1 al 10
Usa un bucle `for` para mostrar los números del 1 al 10, ambos incluidos.

````python
# Escribe tu solución aquí
for cuenta in range(0, 10): # range(inicio, fin)
    cuenta = cuenta + 1
    print(cuenta)
    if cuenta % 2 == 0:
        print("Es par")
````
