## Ejercicio 1. Presentación de una ficha personal

Escribe un programa que muestre una ficha personal con nombre, apellidos, edad y ciclo formativo. Cada dato debe aparecer en una línea diferente y debe haber una tabulación entre la etiqueta y su valor.

**Salida esperada:**

```text
Nombre:        Ana
Apellidos:     García López
Edad:          18
Ciclo:         SMR
```

Utiliza los caracteres de escape adecuados para organizar la información.

````python
# Escribe aquí tu solución al ejercicio 1
print("Nombre:\t\tAna")
print("Apellidos:\tGarcía López")
print("Edad:\t\t18")
print("Ciclo:\t\tSMR")
````
